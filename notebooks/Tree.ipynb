## IMPORT LIBRARY

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/iris-flower.csv', sep=",")
df
```

```python
df.info
```

```python
df['Id'].unique()

```

```python
df['Species'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Species'] = le.fit_transform(df['Species'])
print(df['Species'].unique())
for i, label in enumerate(le.classes_):
    print(f"{i} = {label}")
```

```python
df
```

```python
x= df.drop(['Id', 'Species'], axis=1)
```

```python
x
```

```python
y = df['Species']
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size= 0.25, random_state=0)
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier()
classifier.fit(x_train, y_train)
```

```python
from sklearn import tree

plt.subplots(figsize = (15,10))
tree.plot_tree(classifier, fontsize=10)
plt.show()
```

```python
from scipy.stats import entropy
from sklearn.tree import DecisionTreeClassifier

classifier1 = DecisionTreeClassifier(criterion='entropy', max_depth=5)
classifier1.fit(x_train, y_train)
```

```python
from sklearn import tree

plt.subplots(figsize = (15,10))
tree.plot_tree(classifier1, fontsize=10)
plt.show()
```

```python
depth = classifier.get_depth()
depth
```

```python
from sklearn.metrics import accuracy_score
y_pred = classifier.predict(x_test)
akurasi = accuracy_score(y_test, y_pred)
print(akurasi)
```
